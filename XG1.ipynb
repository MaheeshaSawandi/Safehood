```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder

from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)

import joblib
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
df = pd.read_excel(
    "/content/drive/MyDrive/SafeHood_ML/Dataset/processed_dataset.xlsx"
)

df.head()
```

```python
print(df.shape)

df.info()
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
df["Future_Risk"].value_counts()
```

```python
encoder = LabelEncoder()

df["Location_Code"] = encoder.fit_transform(
    df["Location"]
)
```

```python
features = [

    "Location_Code",

    "Year",

    "Month",

    "Crime_Count",

    "Average_Severity",

    "Maximum_Severity"

]
```

```python
X = df[features]

y = df["Future_Risk"]
```

```python
target_encoder = LabelEncoder()

y_encoded = target_encoder.fit_transform(y)
```

```python
target_encoder.classes_
```

```python
train = df[df["Year"] <= 2023]

test = df[df["Year"] > 2023]
```

```python
X_train = train[features]

X_test = test[features]
```

```python
y_train = target_encoder.transform(
    train["Future_Risk"]
)


y_test = target_encoder.transform(
    test["Future_Risk"]
)
```

```python
print(X_train.shape)

print(X_test.shape)
```

```python
xgb_model = XGBClassifier(

    n_estimators=100,

    learning_rate=0.1,

    max_depth=3,

    random_state=42,

    objective="multi:softmax",

    num_class=3

)
```

```python
xgb_model.fit(

    X_train,

    y_train

)
```

```python
xgb_prediction = xgb_model.predict(

    X_test

)
```

```python
accuracy = accuracy_score(

    y_test,

    xgb_prediction

)

print(
"Accuracy:",
accuracy
)
```

```python
precision = precision_score(

    y_test,

    xgb_prediction,

    average="weighted"

)

print(
"Precision:",
precision
)
```

```python
recall = recall_score(

    y_test,

    xgb_prediction,

    average="weighted"

)

print(
"Recall:",
recall
)
```

```python
f1 = f1_score(

    y_test,

    xgb_prediction,

    average="weighted"

)

print(
"F1 Score:",
f1
)
```

```python
print(
classification_report(

    y_test,

    xgb_prediction,

    target_names=target_encoder.classes_

)
)
```

```python
cm = confusion_matrix(

    y_test,

    xgb_prediction

)


plt.figure(figsize=(6,5))


sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=target_encoder.classes_,

    yticklabels=target_encoder.classes_

)


plt.xlabel(
"Predicted"
)


plt.ylabel(
"Actual"
)


plt.title(
"XGBoost Confusion Matrix"
)


plt.show()
```

```python
probabilities = xgb_model.predict_proba(

    X_test

)
```

```python
importance = pd.DataFrame({

    "Feature": features,

    "Importance": xgb_model.feature_importances_

})


importance = importance.sort_values(

    by="Importance",

    ascending=False

)


importance
```
