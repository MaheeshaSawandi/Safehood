```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder, StandardScaler

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)

import joblib
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
df = pd.read_excel(
    "/content/drive/MyDrive/SafeHood_ML/Dataset/processed_dataset.xlsx"
)

df.head()
```

```python
print(df.shape)

df.info()
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
df["Future_Risk"].value_counts()
```

```python
encoder = LabelEncoder()

df["Location_Code"] = encoder.fit_transform(
    df["Location"]
)
```

```python
features = [

"Location_Code",

"Year",

"Month",

"Crime_Count",

"Average_Severity",

"Maximum_Severity"

]
```

```python
X = df[features]

y = df["Future_Risk"]
```

```python
train = df[df["Year"] <= 2023]


test = df[df["Year"] > 2023]
```

```python
X_train = train[features]

X_test = test[features]


y_train = train["Future_Risk"]

y_test = test["Future_Risk"]
```

```python
scaler = StandardScaler()


X_train_scaled = scaler.fit_transform(
    X_train
)


X_test_scaled = scaler.transform(
    X_test
)
```

```python
lr_01 = LogisticRegression(

    C=0.5,

    max_iter=2000,

    random_state=42

)


lr_01.fit(

    X_train_scaled,

    y_train

)


pred_01 = lr_01.predict(

    X_test_scaled

)
```

```python
print(
"Accuracy:",
accuracy_score(
    y_test,
    pred_01
)
)


print(
"Precision:",
precision_score(
    y_test,
    pred_01,
    average="weighted"
)
)


print(
"Recall:",
recall_score(
    y_test,
    pred_01,
    average="weighted"
)
)


print(
"F1:",
f1_score(
    y_test,
    pred_01,
    average="weighted"
)
)
```

```python
print(
classification_report(
    y_test,
    pred_01
)
)
```

```python
cm = confusion_matrix(
    y_test,
    pred_01
)


plt.figure(figsize=(6,5))


sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=lr_01.classes_,

    yticklabels=lr_01.classes_

)


plt.xlabel(
    "Predicted"
)

plt.ylabel(
    "Actual"
)

plt.title(
    "Logistic Regression Confusion Matrix"
)

plt.show()
```

```python
joblib.dump(

    lr_01,

    "/content/drive/MyDrive/SafeHood_ML/Models/logistic_regression_model2.pkl"

)


joblib.dump(

    scaler,

    "/content/drive/MyDrive/SafeHood_ML/Models/scaler_lr2.pkl"

)
```
