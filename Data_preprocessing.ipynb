```python
severity_map = {

"Abuse":6,

"Kidnapping":10,

"Serious injury":9,

"Assault with Dish":8,

"Murder":10,

"Attempted Murder":10,

"Rape < age 16":10,

"Statutory rape > age 16":9,

"Unnatural offence/Serious sexual abuse":9,

"Cruel treatment of children":9,

"Sexual abuse of children":10,

"Fraud / Human trafficking":9,

"Obstruction of a police officer's duty":4,

"Illegal riot":7,

"Arson / Causing damage with explosive":9,

"Damage > LKR 25,000":3,

"Housebreaking":5,

"The loot":6,

"Forcible acquisition":8,

"Fraud / Breach of trust > LKR 500,000":4,

"Fraud / Breach of trust > LKR 100,000":3,

"Theft of plantation materials > LKR 50,000":2,

"Theft of cows > LKR 50,000":2,

"Theft of property > LKR 50,000":3,

"Printing / Possession of counterfeit notes":6,

"Anti-government activities":8,

"Offenses under the Oppressive Weapons act":8,

"Possession of an automatic / repeater gun":9,

"Keeping Heroin, Cocaine, Morphine production":8

}
```

```python
df["Crime_Severity_Score"] = df["Crime Types"].map(severity_map)
```

```python
monthly = df.groupby(
    [
        "Location",
        "Year",
        "Month"
    ]
).agg(
    Crime_Count=("Crime Types","count"),
    Average_Severity=("Crime_Severity_Score","mean"),
    Maximum_Severity=("Crime_Severity_Score","max")
).reset_index()
```

```python
monthly.head()
```

```python
monthly = monthly.sort_values(
    [
        "Location",
        "Year",
        "Month"
    ]
)
```

```python
monthly["Next_Month_Crime"] = (
    monthly
    .groupby("Location")["Crime_Count"]
    .shift(-1)
)
```

```python
def risk_label(x):

    if x <= 10:
        return "Low"

    elif x <= 30:
        return "Medium"

    else:
        return "High"


monthly["Future_Risk"] = monthly["Next_Month_Crime"].apply(risk_label)
```
